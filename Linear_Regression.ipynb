```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import statsmodels.api as sm
%matplotlib inline
```

```python
data = pd.read_csv(r"C:\Users\fahad\Downloads\Ads.csv")

data.head()
```

```python
data = pd.read_csv("Ads.csv")
```

```python
data.head(10)
```

```python
data.drop(['Unnamed: 0'], axis=1)
```

```python
plt.figure(figsize=(12,6))
plt.scatter(data['TV'], data['sales'], c='green')
plt.xlabel("Money spent on TV ads in Dollars")
plt.ylabel("Sales in Dollars")
plt.show()
```

```python
X = data['TV'].values.reshape(-1,1)
y = data['sales'].values.reshape(-1,1)

reg = LinearRegression()
reg.fit(X, y)
```

```python
#m
print(reg.coef_[0][0])
#b
print(reg.intercept_[0])

print("The linear model is: Y = {:.5} + {:.5}X".format(reg.intercept_[0], reg.coef_[0][0]))
```

```python
predictions = reg.predict(X)

plt.figure(figsize=(10, 7))
plt.scatter( data['TV'], data['sales'], color='blue')
plt.plot( data['TV'], predictions, c='black', linewidth=3)
plt.xlabel("Money spent on TV ads in Dollars")
plt.ylabel("Sales in Dollars")
plt.show()
```

```python
X = data['TV']
y = data['sales']

X2 = sm.add_constant(X)
EST = sm.OLS(y, X2)
EST2 = EST.fit()
print(EST2.summary())
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
from sklearn.metrics import mean_squared_error

# Actual values
y_actual = data['sales']

# Predicted values
y_pred = EST2.predict(X2)

# Evaluation metrics
print("R2 Score:", r2_score(y_actual, y_pred))
print("MAE:", mean_absolute_error(y_actual, y_pred))
print("MSE:", mean_squared_error(y_actual, y_pred))
print("RMSE:", mean_squared_error(y_actual, y_pred)**0.5)
```
